# Reduce

```python
ims = np.load("./stack_nhwc.npy", allow_pickle=False).astype(np.float16)
output = reduce(ims, 'b h w c -> h w c', 'mean')
# Cast to uint8 so PIL can display it properly
display(Image.fromarray(output.astype(np.uint8)))
```

```python
# Add/remove axes
x = rearrange(ims, "b h w c -> b 1 h w 1 c") # somehow related to np.expand_dims?
print(x.shape)
x = rearrange(x, "b 1 h w 1 c -> b h w c") # supposedly like np.squeeze
print(x.shape)
```

```python
x = reduce(ims, "b h w c -> b () () c", "max") - ims
print(x.shape)
# print(x[0][0], x[0][0])
display(Image.fromarray(rearrange(x, "b h w c -> h (b w) c").astype(np.uint8)))
```

# Ultimate Test

```python
list_of_shi = []

for i in range(0, len(ims), 2):
  list_of_shi.append(repeat(ims[i], "h w c -> h w c").astype(np.uint8))

# in-place add axis
tnsr = rearrange(list_of_shi, "b h w c -> b h w c")
print(tnsr.shape)

negatives = reduce(tnsr, "b h w c -> b () () c", "max") - tnsr

repeat(rearrange(negatives, "b h w c -> h (b w) c"), "h w c -> (5 h) (w 2) c")
```
